# [v1 Retrospective] Dataset Analysis
> **What this notebook shows:** Initial inspection of the raw UNSW-NB15 training/testing files, class filtering to retain Normal and 5 major attack categories (Exploits, Fuzzers, DoS, Analysis, Backdoor), and export to data/processed/selected_{training,testing}.csv.
>
> **What went wrong in v1:**
> 1. Retained the raw sequential index column id which leaks scenario timing.
> 2. Did not check for exact duplicate flows between train and test or within train.
> 3. Did not verify conflicting label assignments for identical network flow signatures.

```python

import pandas as pd
import os

# --------------------------------------------------
# 1. Load original datasets
# --------------------------------------------------

train_df = pd.read_csv(
    "../../data/raw/UNSW_NB15_training-set.csv"
)

test_df = pd.read_csv(
    "../../data/raw/UNSW_NB15_testing-set.csv"
)


# --------------------------------------------------
# 2. Count original data points
# --------------------------------------------------

train_count = len(train_df)
test_count = len(test_df)

print("Original Dataset Statistics")
print("--------------------------------")
print(f"Original training records: {train_count}")
print(f"Original testing records: {test_count}")


# --------------------------------------------------
# 3. Select attack categories
# --------------------------------------------------

selected_attacks = [
    "Fuzzers",
    "Analysis",
    "Backdoor",
    "DoS",
    "Exploits"
]


# --------------------------------------------------
# 4. Filter training dataset
# --------------------------------------------------

train_df = train_df[
    (train_df["label"] == 0) |
    (train_df["attack_cat"].isin(selected_attacks))
].copy()


# --------------------------------------------------
# 5. Filter testing dataset
# --------------------------------------------------

test_df = test_df[
    (test_df["label"] == 0) |
    (test_df["attack_cat"].isin(selected_attacks))
].copy()


# --------------------------------------------------
# 6. Count records after filtering
# --------------------------------------------------

train_filtered_count = len(train_df)
test_filtered_count = len(test_df)

print("\nFiltered Dataset Statistics")
print("--------------------------------")
print(f"Filtered training records: {train_filtered_count}")
print(f"Filtered testing records: {test_filtered_count}")

print(
    f"Training records removed: "
    f"{train_count - train_filtered_count}"
)

print(
    f"Testing records removed: "
    f"{test_count - test_filtered_count}"
)


# --------------------------------------------------
# 7. Display category-wise distribution
# --------------------------------------------------

print("\nTraining Category Distribution:")

print(
    train_df["attack_cat"]
    .fillna("Normal")
    .value_counts()
)

print("\nTesting Category Distribution:")

print(
    test_df["attack_cat"]
    .fillna("Normal")
    .value_counts()
)


# --------------------------------------------------
# 9. Save filtered datasets
# --------------------------------------------------

train_df.to_csv(
    "../../data/processed/selected_training.csv",
    index=False
)

test_df.to_csv(
    "../../data/processed/selected_testing.csv",
    index=False
)


# --------------------------------------------------
# 10. Verify saved files
# --------------------------------------------------

print("\nSaved Files:")
print("--------------------------------")

print(
    os.listdir("../../data/processed")
)

print("\nFiltered datasets saved successfully!")
```
